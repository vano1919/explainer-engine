```python
# === CELL 5: Execute Full Unit Test Suite (Including Dramatic-Scene & Cold-Open Cases A–J) ===
import importlib.util, inspect, tempfile, pathlib

suites = [
    "tests/unit/test_anti_robotic_story_first_gates.py",
    "tests/unit/test_cohesive_film_3tier_engine.py",
    "tests/unit/test_d124_gemini_architecture.py",
    "tests/unit/test_dramatic_scene_and_cold_open_integration.py",
    "tests/unit/test_pilot_60s_fixture.py",
    "tests/unit/test_schemas_v4.py",
    "tests/unit/test_shot_prompt_lint_v4.py",
    "tests/unit/test_v4_2_reconciliation_and_viewer_critic.py",
    "tests/unit/test_v4_2_visual_transitions_and_sound_design.py",
]
passed = 0
for p in suites:
    spec = importlib.util.spec_from_file_location(pathlib.Path(p).stem, p)
    mod = importlib.util.module_from_spec(spec)
    spec.loader.exec_module(mod)
    for name, fn in inspect.getmembers(mod, inspect.isfunction):
        if name.startswith("test_"):
            sig = inspect.signature(fn)
            with tempfile.TemporaryDirectory() as td:
                kw = {"tmp_path": pathlib.Path(td)} if "tmp_path" in sig.parameters else {}
                fn(**kw)
                passed += 1
print(f"ALL {passed} UNIT TESTS PASSED ACROSS {len(suites)} SUITES!")
```
