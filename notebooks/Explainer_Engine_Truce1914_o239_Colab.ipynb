```python
# === CELL 1 ($0.00): Mount Google Drive & Unpack Canonical o239-argon-impl Archive ===
import hashlib
import json
import os
import shutil
import zipfile
from pathlib import Path

try:
    from google.colab import drive
    drive.mount('/content/drive')
    DRIVE_ROOT = Path('/content/drive/MyDrive/Explainer_Engine_V4_Main')
except Exception:
    DRIVE_ROOT = Path('.').resolve()

# Canonical archive discovery: prefers canonical/LATEST_COMMIT.json pointer if present,
# otherwise uses REPO_TEXT_o239-argon-impl_e5fa5638.zip + changed_files_handover_e5fa5638.zip.
WORK_DIR = Path('/content/explainer_engine_o239') if Path('/content').exists() else Path('./work_o239').resolve()
if WORK_DIR.exists():
    shutil.rmtree(WORK_DIR)
WORK_DIR.mkdir(parents=True, exist_ok=True)

base_candidates = [
    DRIVE_ROOT / 'canonical' / 'REPO_o239-argon-impl_latest.zip',
    DRIVE_ROOT / 'REPO_TEXT_o239-argon-impl_e5fa5638.zip',
]
for cand in base_candidates:
    if cand.is_file():
        print('Unpacking base repo archive:', cand.name, 'SHA-256:', hashlib.sha256(cand.read_bytes()).hexdigest()[:16])
        with zipfile.ZipFile(cand, 'r') as zf:
            zf.extractall(WORK_DIR)
        break
else:
    raise FileNotFoundError(f'No base repo zip found in {DRIVE_ROOT}')

overlay_zip = DRIVE_ROOT / 'changed_files_handover_e5fa5638.zip'
if overlay_zip.is_file():
    print('Applying handover overlay zip:', overlay_zip.name, 'SHA-256:', hashlib.sha256(overlay_zip.read_bytes()).hexdigest()[:16])
    with zipfile.ZipFile(overlay_zip, 'r') as zf:
        zf.extractall(WORK_DIR)

os.chdir(WORK_DIR)
print('Active working directory:', Path.cwd())
```

```python
# === CELL 2 ($0.00): Schema, Episode Contract, Sourced Rules & Dramatization Audit ===
import json
import subprocess
from pathlib import Path
import jsonschema

schema = json.loads(Path('explainer_engine/schemas/production_script.schema.json').read_text(encoding='utf-8'))
for label, rel_path in [
    ('Variant A', 'episodes/ep_truce_1914/story/production_script.json'),
    ('Variant B', 'episodes/ep_truce_1914/variants/B/production_script.json'),
]:
    doc = json.loads(Path(rel_path).read_text(encoding='utf-8'))
    jsonschema.validate(doc, schema)
    shots = doc['shots']
    assert all(s.get('dramatization_level') in ('documented', 'reconstructed', 'invented') and s.get('basis_claim_ids') for s in shots)
    assert doc.get('docudrama_disclosure', {}).get('include_end_card') is False
    print(f'[{label}] Schema PASS | shots={len(shots)} | visual_mix_target={doc["visual_mix_target"]}')

subprocess.run(['python3', '-m', 'explainer_engine.episode.validate', 'episodes/ep_truce_1914'], check=True)
subprocess.run(['python3', '-m', 'explainer_engine.standards.sourced_rules', 'check'], check=True)
```
